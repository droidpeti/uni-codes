# 1)
A program kérjen egy bemenetet, melyről a felhasználó előre eldönti, hogy az string, vagy integer lesz.

- Ha a bemenet string, akkor írja ki a hosszát, és a benne található "a" és "k" betűk összesített számát.
- Ha a kapott bemenet integeter, akkor adja vissza a 42%-át a 13-al való osztási maradékát és a számjegyeinek összegét.
- És ha a kapott bemenet egyik sem, akkor írja ki, hogy "hibás bemenet".

```python
inp = input("Input: ")
if inp.isdigit():
    num = int(inp)
    sum = 0
    for i in inp:
        sum += int(i)
    print(f"42%: {num*0.42}, mod 13: {num%13}, sum of the digits")
elif inp.isalpha():
    print(f"length: {len(inp)}, occurences of a and k: {inp.lower().count("a")}, {inp.lower().count("k")}")
else:
    print("Bad input")
```

# 2)

Készítsen programot, amely bekér egy x számot, majd elkészíti és eltárolja egy tömbben az első x prímszámot! Oldd meg a feladatot külső modul használata nélkül, majd a <i>sympy</i> megfelelő függvényét használva is!

```python
x = int(input("x="))
t = []
num = 2

while len(t) < x:
    i = 2
    while i < (num**(1/2)) and not num % i == 0:
        i += 1
    prime = i > (num**(1/2))

    if prime:
        t.append(num)
    num += 1

print(t)
```

```python
import sympy

x = int(input("x="))
t = []
num = 2

while len(t) < x:
    if sympy.isprime(num):
        t.append(num)
    num += 1

print(t)
```
