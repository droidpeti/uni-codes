### Kapcsolódás a MySQL adatbázis szerverhez

***pip install mysqlclient***

- Először importáljuk a megfelelő modult:

```python
import MySQLdb
# ha ez nem működik, akkor import mysql.connector
```

- Hozzunk létre egy kapcsolatot:

```python
conn = MySQLdb.connect(
  host="127.0.0.1",
  user="root",
  passwd="root",
  db="iskola",
  port=3306
)
conn
```

Illetve, alternatív esetben ***MysQLdb*** helyett  ***mysql.connector***.

- A ***cursor*** segítségével érjük el az adatbázis szervert az interpreterből:

```python
my_cursor = conn.cursor()
```

- Megnézhetjük, hogy milyen adatbázisok vannak már a rendszerünkben:

```python
my_cursor.execute("SHOW DATABASES")
for i in my_cursor:
    print(i)
```
