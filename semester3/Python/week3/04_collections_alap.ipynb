### 3.3 Halmaz (set) típus

- A matematikában megszokott halmazokat valósítja meg. A megszokott halmazműveletek hatékonyabbak (pl. tagsági viszony eldöntése, fölösleges elemek kigyomlálása), mint listákkal, cserébe rendezetlenek.

- Létrehozás a *set()* függvénnyel, h = {} nem működik.

- A halmazok bejárhatók, támogatja a tagsági műveletet (in), hossz (len()) lekérdezést, összehasonlításokat, bitenkénti műveleteket.

- A **frozenset** a halmaz konstans változata (nem változtatható az értéke). Létrehozása *frozenset()* függvénnyel.

- Halmaz elemei csak **kivonatolható (hashable)** objektumok lehetnek, mint pl. a *float, frozenset, int, str, tuple* típusok. Ezek olyan objektumok, amelyek rendelkeznek a *__hash()__* tagfüggvénnyel, amelynek értéke nem változik az objektum élete során, továbbá más objektumokkal való egyezőségük ellenőrizhatő az *__eq()__* tagfüggvénnyel.

- Tagfüggvények a jegyzet 3.2 táblázatában találhatók.

- **Set comprehension**

   [*kif* for *elem* in *bejárható_objektum*]

   [*kif* for *elem* in *bejárható_objektum* if *feltétel*]

### 3.4 Gyűjtemények bejárása, másolása és törlése

- Vizsgáljuk a különbséget **sekély(shallow)** és **mély(deep)** másolatok közt.

```python
lista_1 = ["4 évszak", "Hawai", "Sonkás", "Frutta del Mare"]
```

**Magyarázat** Azelső esetben sekély másolás történt, azaz a *lista_2* lista *lista_1* memóriacímét kapta meg, fizikai másolat nem jött létre. Ezért a *lista_2*-ben történő értékadás ugyanazt a memóriarészt írja át, ami *lista_1*-hez is tartozik. A második esetben létrejön egy másolat fizikai szinten is, ezért nem érinti *lista_2* változtatása *lista_1*-et. A sekély másolás gyorsabb, hatékonyabb, míg a mély másolás lassabb, költségesebb.   

- Most vizsgéljuk a következő jelenséget:

```python
p = [["4 évszak", "Hawai", "Sonkás", "Frutta del Mare"], "kicsi", "közepes", "nagy", [[7, 9, 5, 12], [10, 12, 8, 15], [12, 14, 10, 17]]]
sp = list(p)
sp[1] = "pici"
p, sp
```

- Nem változott az eredeti lista. Ezek szerint ez egy mély másolat? 

```python
sp[0][0] = "8 évszak"
p, sp
```

- Hupszi! Most megváltozott, tehát sekély, vagy mi? Hát akkor majd szelettel:

```python
p = [["4 évszak", "Hawai", "Sonkás", "Frutta del Mare"], "kicsi", "közepes", "nagy", [[7, 9, 5, 12], [10, 12, 8, 15], [12, 14, 10, 17]]]
sp = p[:]
sp[3] = "óriási"
p, sp
```

- Hát ez már csak jó mély lesz!

```python
sp[0][1] = "Magyaros"
p, sp
```

- Megoldás a **copy modul**:

```python
import copy

p = [["4 évszak", "Hawai", "Sonkás", "Frutta del Mare"], "kicsi", "közepes", "nagy", [[7, 9, 5, 12], [10, 12, 8, 15], [12, 14, 10, 17]]]
sp = copy.deepcopy(p)

sp[3] = "óriási"
sp[0][1] = "Magyaros"
sp[4][0][2] = 100

p, sp
```
