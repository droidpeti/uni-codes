### Tengelycímkék és szöveg

```python
mu, sigma = 115, 15
x = mu + sigma * np.random.randn(10000)
```

```python
fig, ax = plt.subplots()
ax.hist(x, 50, density=True)

ax.set_xlabel("Pizzas")
ax.set_ylabel("Time")
ax.set_title("Pizza Shop")
ax.grid(True)
ax.text(75, 0.025, r"$\mu=115,\ \sigma=15$")
```

Minden szövegfüggvény egy matplotlib.text.Text.Text példányt ad vissza.

```python
fig, ax = plt.subplots()
n, bins, patches = ax.hist(x, 50, density=True)

t = ax.set_xlabel('X Label', fontsize=14, color='red')
ax.set_ylabel('Y Label')
ax.set_title('Title\n Sub Title')
ax.text(75, .025, r'$\mu=115,\ \sigma=15$')
ax.grid(True)

print(type(t))
```

## Artist tulajdonságainak módosítása

```python
x = np.arange(0, 13, 0.2)
y = np.sin(x)

fig, ax = plt.subplots(figsize=(4, 2.5))
lines = ax.plot(x, y, '-', label='example', linewidth=0.2, color='blue')
#lines[0].set(color='green', linewidth=2)
```

## Artist adatainak módosítása

```python
x = np.arange(0, 13, 0.2)
y = np.sin(x)

fig, ax = plt.subplots(figsize=(4, 2.5))
lines = ax.plot(x, y, '-', label='example')
lines[0].set_data([x, np.cos(x)])
```

## Stílus

# Haladó

## Testreszabott Colorbar-ok

```python
fig, ax = plt.subplots(figsize=(6, 1))
fig.subplots_adjust(bottom=0.5)

cmap = mpl.cm.cool
norm = mpl.colors.Normalize(vmin=5, vmax=10)

fig.colorbar(mpl.cm.ScalarMappable(norm=norm, cmap=cmap), cax=ax, orientation='horizontal', label='Some Units')
```

## Kibővített Colorbar folyamatos színskálával

```python
cmap = mpl.cm.viridis
bounds = [-1, 2, 5, 7, 12, 15]
norm = mpl.colors.BoundaryNorm(bounds, cmap.N, extend='both')

fig, ax = plt.subplots(figsize=(6, 1))
fig.subplots_adjust(bottom=0.5)

fig.colorbar(mpl.cm.ScalarMappable(norm=norm, cmap=cmap),
             cax=ax, orientation='horizontal',
             label="Discrete intervals with extend='both' keyword")
```

## Szöveg

### Szöveg a Matplotlib ábrákon

```python
fig = plt.figure()
ax = fig.add_subplot()
fig.subplots_adjust(top=0.85)

# Set titles for the figure and the subplot respectively
fig.suptitle('bold figure suptitle', fontsize=14, fontweight='bold')
ax.set_title('axes title')

ax.set_xlabel('xlabel')
ax.set_ylabel('ylabel')

# Set both x- and y-axis limits to [0, 10] instead of default [0, 1]
ax.axis([0, 10, 0, 10])

ax.text(3, 8, 'boxed italics text in data coords', style='italic',
        bbox={'facecolor': 'red', 'alpha': 0.5, 'pad': 10})

ax.text(2, 6, r'an equation: $E=mc^2$', fontsize=15)

ax.text(3, 2, 'Unicode: Institut für Festkörperphysik')

ax.text(0.95, 0.01, 'colored text in axes coords',
        verticalalignment='bottom', horizontalalignment='right',
        transform=ax.transAxes,
        color='green', fontsize=15)

ax.plot([2], [1], 'o')
ax.annotate('annotate', xy=(2, 1), xytext=(3, 4),
            arrowprops=dict(facecolor='black', shrink=0.05))

plt.show()
```

## scatterplot

```python
data = {'a': np.arange(50),
        'c': np.random.randint(0, 50, 50),
        'd': np.random.randn(50)}
data['b'] = data['a'] + 10 * np.random.randn(50)
data['d'] = np.abs(data['d']) * 100

fig, ax = plt.subplots(figsize=(5, 2.7), layout='constrained')
ax.scatter('a', 'b', c='c', s='d', data=data)
ax.set_xlabel('entry a')
ax.set_ylabel('entry b')
```
