**1.2.1.2 feladat.** Írjunk programot, amely kiírja adott szám faktoriálisát és ennek hosszát (decimális jegyeinek számát). 

```python
num = int(input("Enter a natural number: "))

if num < 0:
    print("Stupid")
else:
    fact = 1;
    for i in range(1, num+1):
        fact *= i

print(f"{num}! = {fact}") 
```

**1.2.1.3 feladat.** Írjunk meg a faktoriális számoló rutint rekurzív függvénnyel.

```python
def fact(num):
    if num <= 1:
        return 1
    return num * fact(num-1)

num = int(input("Enter a natural number: "))

print(f"{num}! = {fact(num)}") 
```

**1.2.1.4 feladat.** Írjunk rutint, amely adott szövegben visszaadja egy adott betű előfordulásainak a számát.

**1.2.1.5 feladat.** Implementáljunk egy fordított bináris számkitalálós játékot. 

```python
import random

n = int(input("Enter a natural number! "))
round = 0

guess = random.randrange(1,101)
while guess != n:
    if guess > n:
        print(f"Too big {guess}")
    elif guess < n:
        print(f"Too small {guess}")
    guess = random.randrange(1,101)
    round += 1

print(f"You guessed it in {round} rounds!")
```

### 1.2.4 Logikai kifejezések

**is** azonossági művelet, összehasonlító műveletek pl. **==, <, >, !=, <=, >=**.

```python
ez = "alma"
az = "alma"
print(ez, az)
print(ez is az)
ez = "alma", "körte"
az = "alma", "körte"
print(ez, az)
print(ez is az)
print(ez == az)
```

**1.2.4.1 feladat** Magyarázzuk meg, hogy az első **is** utasításnak miért *True* az értéke, a másiknak pedig *False*, aztán megint azt kapjuk, hogy ***ez*** egyenlő ***az***. 

Mindkét esetben a változók értéke ugyanaz, viszont második esetben a tupletek objektumok, így létrejöttükkor más memóriacímre kerültek. Az **==** művelet viszont értékeket hasonlít össze, ezért lesz igaz az utolsó eredmény.

- Az **is** egy észszerű használata: van-e értéke egy adott változónak.

- Az összehasonlító utasítások összeláncolhatók.

Az **in** tagsági művelet. Igen hasznos, de mivel lineáris keresést végez, ezért csak kis méretű gyűjteményekben használatos, vagy szótár, illetve halmaz adatszerkezetnél.  

#### Logikai műveletek nem logikai típusú értékekkel

- Logikai **False**-nak értékelődnek ki a **None**, bármely típusú **0** és bármely **üres adatszerkezet** (string, tuplet, lista, szótár és halmazok).

- Minden egyéb beépített érték **True**-nak értékelődik ki.

- A kiértékelés a logikai megfelelők szerint történik, de a visszaadott érték az "eredeti".

- Az *x* **and** *y* utasításnál először *x* értékelődik ki, ha *x* hamis, akkor *x* értéke adódik vissza, ha igaz, akkor *y*-é.

- Az *x* **or** *y* utasításnál először *x* értékelődik ki, ha *x* igaz, akkor *x* értéke adódik vissza, ha nem, akkor *y*-é.
