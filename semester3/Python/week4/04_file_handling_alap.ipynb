# Fájlműveletek: $\textit{open(), with}$

Az $\textit{open()}$ egy beépített fájlművelet, nem szükséges hozzá importálni semmit.

Hivatalos dokumentáció: $\href{https://docs.python.org/3/library/functions.html?highlight=open#open}{itt}$

Minimális szintaktika: $\textbf{open(file)}$

Teljes szintaktika: open(file, mode='r', buffering=-1, encoding=None, errors=None, newline=None, closefd=True, opener=None)

Egy $\textit{file}$ objektumot kapunk vissza. (Bővebben a $\textit{file}$ obejektumokról $\href{https://docs.python.org/3/glossary.html#term-file-object}{itt}$.)

$\large{Open("file","*")}$:

- file: a fájl elérési útja (abszolút, vagy relatív)

- *: a fájl megnyitásának módja

$\textit{Open("file","r")}$: olvasás, alapértelmezett.

```python
file_name = "input.txt"
```

$\textit{Open(file,w)}$: írás.<br> Ha a fájl nem létezik, akkor létrehozza. Ha létezik, akkor felülírja.

```python
file = open(file_name, "r")

print(file.read())

print(file.readline())

file.close()
```

$\large{\textit{with}}$: elvégzi a fájl megnyitását és bezárását, ha végzett.

$\textit{Open("file","x")}$: "exkluzív" írás.<br> Ha a fájl létezik, akkor hibával (FileExistsError) kilép.

$\textit{Open("file","a")}$: megnyitás, hozzáfűzés.<br> Mindig a fájl végéhez.

Itt is alkalmazható a $\textit{with}$:

$\textit{Open("file","*t")}$: text mód (alapértelmezett)<br> Elhagyható, mivel alapból így nyitja meg a fájlokat a Python.

$\textit{*t}$: * helyére "r", "w", "a" vagy "x" kerülhet, önmagában nem működik.

$\textit{Open("file","*b")}$: bináris mód<br> A fájl tartalmát $\textit{bájtként}$ jeleníti meg, kódolás nélkül.

$\textit{*b}$: * helyére "r", "w", "a", vagy "x" kerül, önmagában a "b" mód sem működik a "t"-hez hasonlóan.

Latin - 2 kódolás, ami tartalmazza az összes magyar karaktert: $\textbf{iso8859-15}$

$\textit{Open("file","*+**")}$: fájl "frissítése"<br> "*" helyére jön "r", "w", stb., "**" helyére pedig "t" vagy "b".

Megjegyzés: Ez esetben "r" módban is lehet $\textit{.write()}$ parancsot kiadni.<br>
A $\textit{.write()}$ ebben a "+" módban felülírja a szöveget az elejétől az új szöveggel.<br><br>
Példa:<br>
    Eredeti szöveg:<br>
        Ez az eredeti szöveg, amit felül fogunk írni.<br><br>
    Megnyitjuk a fájlt, majd "beleírjuk", hogy "bükkfa kapitány".<br>
    Eredmény:<br>
        bükkfa kapitányzöveg, amit felül fogunk írni.
    

Megjegyzés: Ha egy fájlhoz hozzáadunk egy kódolt adatot, akkor az egész fájl kódolása megváltozik.
