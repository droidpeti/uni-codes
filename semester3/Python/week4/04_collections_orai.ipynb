**3.2.2 Feladat** Elemezzük a következő hibát és orvosoljuk a kicsomagolás művelettel. Magyarázzuk meg a történéseket.

```python
def konkat(x, y, z):
  return x + y + z

lista = ["Boldi ", "elment ", "pizzát enni."]
# print(konkat(lista)) -> nem jó
# Ctrl + K + C <-> Ctrl + K + U
print(konkat(*lista)) # -> ez már jó
```

- Első esetben azért kaptunk hibát, mert a *konkat* függvény 3 paramétert vár, de csak 1-et kap. Második esetben már jó, mivel a kicsomagoló művelet 3 elemre vágja az átadott listát.

- **Szeletelés** és **egyelés**, olyan műveletek, amelyekkel nagyon hatékonyan tudunk lista darabokra hivatkozni, illetve listát bejárni, módosítani.

```python
szamok = [23, 356, 3612, 353, 735]
print(szamok[1:3]) # -> 1. indextől 3-1 elem

del szamok[1:3]
print(szamok) # 23 353 735

szamok.insert(3, 100)
print(szamok) # 23 353 735 100

szamok[1:3] = [21]
print(szamok) # 23 21 100
```

**3.2.3 Feladat** Készítsünk egy konzekutív, 1 és *n* közti, 3-mal osztva 1-et adó egészeket tartalmazó listát.

```python
n = 100
p = [x for x in range(1, n+1) if x%3==1]
print(p)

p = []
for x in range(1, n+1): # [... for x in range(1, n+1) ...]
  if x%3==1: # [.. if x%3==1]
    p.append(x) # [x for ...]
print(p)
```

- **List comprehension (listaelemek előállítása)**

   [*elem* for *elem* in *bejárható_objektum*]

   [*kif* for *elem* in *bejárható_objektum*]

   [*kif* for *elem* in *bejárható_objektum* if *feltétel*]

**3.2.4 Feladat** Egy konzekutív egészeket tartalmazó listát szitáljunk ki 2-vel (párosokat írjuk át 0-ra).

```python
n = 30
szamok = [x for x in range(1, n+1)]
print(szamok)
szamok[1::2] = [0] * len(szamok[1::2]) # intervallum = mire * hányat
print(szamok)
del szamok[::2] # elejétől, 2-esével töröl elemeket
print(szamok)
```

### 3.3 Halmaz (set) típus

- A matematikában megszokott halmazokat valósítja meg. A megszokott halmazműveletek hatékonyabbak (pl. tagsági viszony eldöntése, fölösleges elemek kigyomlálása), mint listákkal, cserébe rendezetlenek.

- Létrehozás a *set()* függvénnyel, h = {} nem működik.

- A halmazok bejárhatók, támogatja a tagsági műveletet (in), hossz (len()) lekérdezést, összehasonlításokat, bitenkénti műveleteket.

- A **frozenset** a halmaz konstans változata (nem változtatható az értéke). Létrehozása *frozenset()* függvénnyel.

- Halmaz elemei csak **kivonatolható (hashable)** objektumok lehetnek, mint pl. a *float, frozenset, int, str, tuple* típusok. Ezek olyan objektumok, amelyek rendelkeznek a *__hash()__* tagfüggvénnyel, amelynek értéke nem változik az objektum élete során, továbbá más objektumokkal való egyezőségük ellenőrizhatő az *__eq()__* tagfüggvénnyel.

- **Set comprehension**

   [*kif* for *elem* in *bejárható_objektum*]

   [*kif* for *elem* in *bejárható_objektum* if *feltétel*]

**3.3.1 Feladat** Boldi mindig eltárolja egy listában, hogy milyen pizzát evett aznap. Írjuk ki gyorsan, hány különbözőt evett eddig.

```python
import random
p = [random.randint(1, 10) for i in range(1, 6)]
print("Pizzák:", p)
print(set(p))
print("10 különböző pizzából 5 evéssel Boldi", len(set(p)) ," fajtát kóstolt meg.")
```

### 3.4 Szótár (dictionary) típus

- Mint a halmaz, csak "kulcs : érték" párok az elemei. Alaptípusok ***dict()*** és ***collections.defaultdict()***.

- Egy szótár kulcsai csak **kivonatolható (hashable)** objektumok lehetnek, mint pl. a *float, frozenset, int, str, tuple* típusok, az értékek bármilyen típusúak.

- Létrehozás a *dict()* függvénnyel vagy ={} felírással.

```python
d1 = dict({"id": 1, "name": "Tupli", "price": 5})
d2 = dict(id=2, name="Hawaii", price=3)
d3 = dict([("id", 3), ("name", "Songoku"), ("price", 3)])
d4 = dict(zip(("id", "name", "price"), (4, "Napolyi", 10)))
d5 = {"id": 5, "name": "Sajtos", "price": 6}

d1, d2, d3, d4, d5

print(d2["price"])

d1["name"] = "Maki"
print(d1["name"])

del d5["id"]
print(d5)
```

```python
print("Elemek:")
for i in d1.items():
  print(i)

print("\nKulcsok:")
for i in d1.keys(): # <-> d1
  print(i)

print("\nÉrtékek:")
for i in d1.values():
  print(i)
```

A *zip()* függvény működése.

- A szótárak bejárhatók (a bejárás sorrendje megjósolhatatlan), támogatja a tagsági műveletet (in), hossz (len()) lekérdezést, bizonyos összehasonlításokat.

- Beszúrás, módosítás, törlés, bejárás (kulcs és érték szerint is lehet).

- **Dictionary comprehension**

   [*kulcs_kif : érték_kif* for *kulcs, érték* in *bejárható_objektum*]

   [*kulcs_kif : érték_kif* for *kulcs, érték* in *bejárható_objektum* if *feltétel*]
