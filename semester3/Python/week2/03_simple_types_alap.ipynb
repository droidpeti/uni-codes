**2.1.1 Feladat** Írjunk programot, amely egy megadott intervallumban keres 2-es alapú páros pszeudoprímeket. Hibakezelést oldjuk meg, az ellenőrzés az intervallum első páros számánál kezdődjön.

n pszeudoprím/álprím, ha 2^n - 1 ≡ 1 (n) és n összetett páros szám.

```python
#1

import time

print("Be: a és b, ahol 2 < a < b egészek.")

try:
    a = int(input("a: "))
    b = int(input("b: "))
    count = 0
    start = time.time()

    if 2 < a < b:
        for n in range(a+(a%2),b,2):
            if 2**n - 2 % n == 0:
                print(n)
                count += 1
        if count == 0:
            print("No pseudo primes found")
        print(f"Passed time: {time.time() - start}s")
    else:
        print("bad interval")
except:
    print("Stupid")
```
